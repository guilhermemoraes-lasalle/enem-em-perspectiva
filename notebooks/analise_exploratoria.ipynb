# Exploração dos agregados do Enem 2025

Este notebook lê somente as tabelas agregadas do projeto. A análise dos registros individuais é feita em blocos por `src/etl.py`; identificadores pessoais não são carregados aqui.

```python
from pathlib import Path
import pandas as pd

resumo = pd.read_csv(Path('../data/processed/resumo_escola_area.csv'))
controle = pd.read_csv(Path('../data/processed/controle_processamento.csv'))
display(controle)
display(resumo.head(10))
```

```python
# Médias ponderadas por quantidade de notas válidas, por rede e área.
resumo.groupby(['dependencia_administrativa', 'area']).apply(
    lambda g: pd.Series({
        'n_notas_validas': g['n_notas_validas'].sum(),
        'media_ponderada': (g['media'] * g['n_notas_validas']).sum() / g['n_notas_validas'].sum()
    }), include_groups=False
).round(1).reset_index()
```

As comparações são descritivas. Médias agregadas não ajustam diferenças na composição dos participantes ou das escolas.
